```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

player = pd.read_csv('players_22.csv')
print(player.shape)
player.head(10)
```

```python
player.info()
print()
print("missing values per column")
print(player.isnull().sum())  # isnull().sum() - number of missing values in each column
```

```python
# Drop rows where value_eur is missing
player = player.dropna(subset=["value_eur"]).reset_index(drop=True)

# Keep only numeric columns
Bank = player.select_dtypes(include="number").copy()

extra_columns = ["sofifa_id", "club_team_id", "nationality_id", "club_jersey_number",
                 "nation_team_id", "nation_jersey_number", "release_clause_eur"]
Bank = Bank.drop(columns=[c for c in extra_columns if c in Bank.columns])

Bank = Bank.fillna(Bank.median())

print("Remaining missing values:", Bank.isnull().sum().sum())  # should be 0
print(Bank.shape)
Bank.head()
```

```python
from sklearn.model_selection import train_test_split

X = Bank.drop(columns=["value_eur"])
y = Bank["value_eur"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.svm import SVR

models = {
    "Decision Tree":           DecisionTreeRegressor(random_state=42),
    "KNN (k-nearest neighbors)": KNeighborsRegressor(),
    "Random Forest":           RandomForestRegressor(random_state=42),
    "Gradient Boosting":       GradientBoostingRegressor(random_state=42),
    "Ridge Regression":        Ridge(),
    "SVR":                     SVR(),
}
```

```python
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    results[name] = round(model.score(X_test, y_test) * 100, 1)

print(results)
```

```python
model_names = list(results.keys())
model_scores = list(results.values())

plt.figure(figsize=(8, 4))
plt.barh(model_names, model_scores, color="mediumseagreen")
plt.xlabel("score (%)")
plt.title("Comparison of 6 models (predicting player value)")
plt.tight_layout()
plt.show()
```

```python
cluster_data = Bank[["overall", "value_eur"]]
cluster_data.head()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X_cluster = StandardScaler().fit_transform(cluster_data)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
Bank["Group"] = kmeans.fit_predict(X_cluster)

Bank[["overall", "value_eur", "Group"]].head(500)
```

```python
from sklearn.preprocessing import StandardScaler

# Sort the groups by average rating and name them (lowest -> highest)
order = Bank.groupby("Group")["overall"].mean().sort_values().index
tier_names = ["Regular players", "Mid-level players",
              "Strong players", "Star players (TOP)"]
tier_map = dict(zip(order, tier_names))

scaler = StandardScaler().fit(cluster_data)

ronaldo_2008_prime = pd.DataFrame({"overall": [100], "value_eur": [100000000]})
group = kmeans.predict(scaler.transform(ronaldo_2008_prime))[0]
print("Ronaldo 2008 Prime:  ", tier_map[group])
```
